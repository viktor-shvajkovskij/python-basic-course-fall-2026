1.  **Калькулятор ИМТ (Индекс Массы Тела):** Напишите функцию, которая принимает вес (в кг) и рост (в метрах) и возвращает ИМТ. Формула: `вес / (рост * рост)`.

```python
def calculate_bmi(weight: float, height: float) -> float:
    return weight / (height ** 2)

user_weight = float(input("Введите ваш вес в кг: "))
user_height = float(input("Введите ваш рост в метрах (например, 1.75): "))
bmi = calculate_bmi(user_weight, user_height)
print(f"Ваш ИМТ: {bmi:.2f}")
```

2. **Расчет стоимости с налогом:** Напишите функцию `calculate_total_price`, которая принимает цену товара и процент налога (например, 20 для 20%). Функция должна возвращать итоговую стоимость товара, включая налог.

```python
def calculate_total_price(price: float, tax_rate: float) -> float:
    return price + (price * tax_rate / 100)

item_price = float(input("Введите цену товара: "))
item_tax = float(input("Введите процент налога (например, 20): "))

total_price = calculate_total_price(item_price, item_tax)
print(f"Итоговая стоимость с налогом: {total_price:.2f}")
```

3.  **Конвертер валют:** Напишите функцию, которая принимает сумму в долларах и курс обмена, а возвращает сумму в рублях.

```python
def convert_usd_to_rub(usd_amount: float, exchange_rate: float) -> float:
    return usd_amount * exchange_rate

dollars = float(input("Введите сумму в долларах ($): "))
rate = float(input("Введите текущий курс доллара к рублю: "))

rubles = convert_usd_to_rub(dollars, rate)
print(f"Сумма в рублях: {rubles:.2f} руб.")
```

4.  **Площадь и периметр:** Создайте функцию, которая принимает длину и ширину прямоугольника и возвращает **два** значения: его площадь и периметр.

```python
def calculate_rectangle_properties(length: float, width: float) -> tuple[float, float]:
    area = length * width
    perimeter = 2 * (length + width)
    return area, perimeter

rect_length = float(input("Введите длину прямоугольника: "))
rect_width = float(input("Введите ширину прямоугольника: "))

area, perimeter = calculate_rectangle_properties(rect_length, rect_width)

print(f"Площадь: {area:.2f}")
print(f"Периметр: {perimeter:.2f}")
```

5.  **Простой процент:** Напишите функцию `calculate_deposit`, которая принимает начальную сумму вклада, годовой процент и количество лет, а возвращает итоговую сумму (без учета капитализации).

```python
def calculate_deposit(principal: float, annual_rate: float, years: int) -> float:
    interest = principal * (annual_rate / 100) * years
    return principal + interest

deposit_amount = float(input("Введите начальную сумму вклада: "))
rate = float(input("Введите годовую процентную ставку (%): "))
deposit_years = int(input("Введите срок вклада (в годах): "))

total_amount = calculate_deposit(deposit_amount, rate, deposit_years)
print(f"Итоговая сумма вклада через {deposit_years} лет: {total_amount:.2f}")
```

7.  **Конвертер времени:** Напишите функцию, которая принимает количество часов и возвращает, сколько это будет в минутах и секундах (два значения).

```python
def convert_hours_to_minutes_and_seconds(hours: float) -> tuple[float, float]:
    minutes = hours * 60
    seconds = hours * 3600
    return minutes, seconds

input_hours = float(input("Введите количество часов: "))
mins, secs = convert_hours_to_minutes_and_seconds(input_hours)

print(f"Минут: {mins:.2f}")
print(f"Секунд: {secs:.2f}")
```

8.  **Расчет гипотенузы:** Создайте функцию, принимающую два катета прямоугольного треугольника и возвращающую его гипотенузу по теореме Пифагора.

```python
def calculate_hypotenuse(a: float, b: float) -> float:
    return (a ** 2 + b ** 2) ** 0.5

leg_a = float(input("Введите длину первого катета: "))
leg_b = float(input("Введите длину второго катета: "))

hypotenuse = calculate_hypotenuse(leg_a, leg_b)
print(f"Длина гипотенузы: {hypotenuse:.2f}")
```

9.  **Следующее и предыдущее:** Напишите функцию, которая принимает одно целое число и возвращает два числа: предыдущее и следующее.

```python
def get_adjacent_numbers(number: int) -> tuple[int, int]:
    previous_num = number - 1
    next_num = number + 1
    return previous_num, next_num

user_number = int(input("Введите целое число: "))

prev_n, next_n = get_adjacent_numbers(user_number)

print(f"Предыдущее число: {prev_n}")
print(f"Следующее число: {next_n}")
```

10. **Стоимость поездки:** Создайте функцию, которая принимает расстояние (км), расход топлива (л/100км) и цену за литр, а возвращает общую стоимость поездки.

```python
def calculate_trip_cost(distance: float, consumption: float, price_per_liter: float) -> float:
    total_liters = (distance / 100) * consumption
    return total_liters * price_per_liter

trip_distance = float(input("Введите расстояние поездки (км): "))
fuel_consumption = float(input("Введите расход топлива (л/100 км): "))
fuel_price = float(input("Введите цену за 1 литр топлива: "))

total_cost = calculate_trip_cost(trip_distance, fuel_consumption, fuel_price)
print(f"Общая стоимость поездки: {total_cost:.2f} руб.")
```
