**Задача 1: Анализатор идентичности**

Напишите функцию `analyze_identity(a, b)`, которая возвращает одну из следующих строк:
*   `"Полная идентичность"` (если `a` и `b` — это один и тот же объект).
*   `"Равенство по значению"` (если это разные объекты, но их значения равны).
*   `"Полное различие"` (если и объекты, и их значения разные).

```python
import ast

def analyze_identity(a, b):
    if a is b:
        return "Полная идентичность"
    elif a == b:
        return "Равенство по значению"
    else:
        return "Полное различие"

def parse_input(prompt):
    raw = input(prompt)
    try:
        return ast.literal_eval(raw)
    except (ValueError, SyntaxError):
        return raw

val_a = parse_input("Введите первое значение: ")
val_b = parse_input("Введите второе значение: ")

print(analyze_identity(val_a, val_b))
```

**Задача 3: Нормализатор и валидатор email**

Напишите функцию `validate_email(email_string)`, которая возвращает `True` или `False`. Чтобы email считался валидным, он должен (после очистки от крайних пробелов и приведения к нижнему регистру) удовлетворять трем условиям:
1.  Содержать ровно один символ `'@'`.
2.  Символ `'@'` не может быть первым или последним символом.
3.  Часть после `'@'` (домен) должна содержать хотя бы одну точку, которая не является первым или последним символом домена.


```python
def validate_email(email_string: str) -> bool:
    cleaned = email_string.strip().lower()
    
    if cleaned.count('@') != 1:
        return False
    
    local, domain = cleaned.split('@')
    
    if not local or not domain:
        return False
    
    if '.' not in domain:
        return False
    
    if domain.startswith('.') or domain.endswith('.'):
        return False
    
    return True

user_email = input("Введите email для проверки: ")
result = validate_email(user_email)

print(f"Валидность email: {result}")
```

**Задача 4: Анализ "бутылочного горлышка"**

Напишите функцию `analyze_concatenation(n)`, которая не выполняет сложение строк, а возвращает строку-отчет. Отчет должен описывать, сколько **всего** промежуточных строковых объектов будет создано в памяти (и затем удалено сборщиком мусора), если выполнить цикл конкатенации `n` раз (как в загадке из лекции).
*Пример вызова:* `analyze_concatenation(3)` должен вернуть что-то вроде `"Для 3 итераций будет создано 3 промежуточных объекта ('a', 'ab', 'abc')."`


```python
def analyze_concatenation(n: int) -> str:
    if n <= 0:
        return "Количество итераций должно быть больше 0."
    
    samples = []
    current_str = ""
    for i in range(n):
        current_str += chr(ord('a') + (i % 26))
        samples.append(f"'{current_str}'")
    
    samples_text = ", ".join(samples)
    return f"Для {n} итераций будет создано {n} промежуточных объектов ({samples_text})."

n_input = int(input("Введите количество итераций (n): "))
print(analyze_concatenation(n_input))
```

**Задача 5: Детектор кодировок**

Напишите функцию `check_encoding_decoding(text)`, которая принимает строку с кириллицей. Функция должна:
1.  Закодировать эту строку в `utf-8` и в `windows-1251`.
2.  Попытаться декодировать байты `windows-1251` обратно в строку, используя `utf-8`.
3.  Вернуть кортеж из трех элементов: исходный текст, корректно декодированный текст и некорректно декодированный текст.


```python
def check_encoding_decoding(text: str) -> tuple:
    encoded_utf8 = text.encode('utf-8')
    encoded_win1251 = text.encode('windows-1251')
    
    correct_text = encoded_win1251.decode('windows-1251')
    incorrect_text = encoded_win1251.decode('utf-8', errors='replace')
    
    return (text, correct_text, incorrect_text)

user_text = input("Введите текст на кириллице: ")
original, correct, corrupted = check_encoding_decoding(user_text)

print(f"Исходный текст: {original}")
print(f"Корректный текст: {correct}")
print(f"Искаженный текст: {corrupted}")
```

**Задача 6: Парсер команд робота**

Напишите функцию `parse_robot_command(command)`, которая принимает одну строку. Используя `match/case`, разберите команду:
*   Команды `"вперед"` или `"назад"` должны возвращать `"Робот движется..."`.
*   Команды `"налево"` или `"направо"` должны возвращать `"Робот поворачивает..."`.
*   Команда `"стоп"` должна возвращать `"Робот остановлен."`.
*   Любая другая команда должна возвращать `"Неизвестная команда."`.

Используйте оператор `|` для группировки похожих команд.


```python
def parse_robot_command(command: str) -> str:
    cmd = command.strip().lower()
    
    match cmd:
        case "вперед" | "назад":
            return "Робот движется..."
        case "налево" | "направо":
            return "Робот поворачивает..."
        case "стоп":
            return "Робот остановлен."
        case _:
            return "Неизвестная команда."

user_cmd = input("Введите команду для робота: ")
print(parse_robot_command(user_cmd))
```

**Задача 7: Валидатор PIN-кода**

Напишите функцию `is_valid_pin(pin_code)`, которая проверяет, что строка с PIN-кодом состоит **только** из цифр. Длина PIN-кода должна быть ровно 4 или 6 символов. Функция должна вернуть `True` или `False`.
**Обязательное условие:** Для проверки на наличие не-цифровых символов используйте цикл `for` с блоком `else`.


```python
def is_valid_pin(pin_code: str) -> bool:
    if len(pin_code) not in (4, 6):
        return False
    
    for char in pin_code:
        if not char.isdigit():
            break  # Нарушаем цикл, если нашли не-цифру
    else:
        return True
    
    return False

user_pin = input("Введите PIN-код: ")
print(f"Валидность PIN-кода: {is_valid_pin(user_pin)}")
```

**Задача 8: Фильтр "стоп-символов"**

Напишите функцию `filter_text(text, stop_chars)`, которая принимает основной текст и строку "стоп-символов". Функция должна вернуть новую строку, в которой отсутствуют все символы, перечисленные в `stop_chars`.
**Обязательное условие:** Используйте цикл `for` с оператором `continue`.


```python
def filter_text(text: str, stop_chars: str) -> str:
    result = []
    
    for char in text:
        if char in stop_chars:
            continue
        result.append(char)
        
    return "".join(result)

user_text = input("Введите текст: ")
user_stops = input("Введите стоп-символы: ")

filtered = filter_text(user_text, user_stops)
print(f"Результат фильтрации: {filtered}")
```

**Задача 9: Шифр Цезаря с `enumerate`**

Напишите функцию `caesar_cipher_encoder(text)`, которая "шифрует" текст по следующему правилу: символы на четных индексах сдвигаются на 1 вперед по алфавиту (`a`->`b`, `b`->`c`), а символы на нечетных индексах сдвигаются на 1 назад (`c`->`b`, `b`->`a`). Пробелы и знаки препинания не меняются.
**Обязательное условие:** Используйте `enumerate` для получения индекса и символа в цикле.
*(Подсказка: функции `ord()` и `chr()` могут быть полезны для работы с кодами символов).*

*(Дополнительное условие: шифр должен корректно обрабатывать "зацикливание" алфавита (например, `z` на четной позиции должна становиться `a`, а `a` на нечетной — `z`). Также шифр должен сохранять регистр букв: заглавные буквы шифруются в заглавные, а строчные — в строчные).*


```python
def caesar_cipher_encoder(text: str) -> str:
    result = []
    
    for idx, char in enumerate(text):
        if 'a' <= char <= 'z':
            shift = 1 if idx % 2 == 0 else -1
            new_char = chr(ord('a') + (ord(char) - ord('a') + shift) % 26)
            result.append(new_char)
        elif 'A' <= char <= 'Z':
            shift = 1 if idx % 2 == 0 else -1
            new_char = chr(ord('A') + (ord(char) - ord('A') + shift) % 26)
            result.append(new_char)
        elif 'а' <= char <= 'я':
            shift = 1 if idx % 2 == 0 else -1
            new_char = chr(ord('а') + (ord(char) - ord('а') + shift) % 32)
            result.append(new_char)
        elif 'А' <= char <= 'Я':
            shift = 1 if idx % 2 == 0 else -1
            new_char = chr(ord('А') + (ord(char) - ord('А') + shift) % 32)
            result.append(new_char)
        else:
            result.append(char)
            
    return "".join(result)

user_text = input("Введите текст: ")
print(f"Зашифрованный текст: {caesar_cipher_encoder(user_text)}")
```

**Задача 10: Комплексный валидатор пароля**

Напишите функцию `is_strong_password(password)`, которая проверяет пароль на надежность и возвращает `True` или `False`.
Надежный пароль должен соответствовать **всем** следующим критериям:
1.  Длина не менее 10 символов.
2.  Содержит хотя бы одну букву в верхнем регистре (`A-Z`).
3.  Содержит хотя бы одну букву в нижнем регистре (`a-z`).
4.  Содержит хотя бы одну цифру (`0-9`).
5.  Содержит хотя бы один из спецсимволов: `!@#$%^&*()`.


```python
def is_strong_password(password: str) -> bool:
    if len(password) < 10:
        return False
    
    special_characters = "!@#$%^&*()"
    
    has_upper = False
    has_lower = False
    has_digit = False
    has_special = False
    
    for char in password:
        if char.isupper():
            has_upper = True
        elif char.islower():
            has_lower = True
        elif char.isdigit():
            has_digit = True
        elif char in special_characters:
            has_special = True
            
    return has_upper and has_lower and has_digit and has_special

user_password = input("Введите пароль для проверки: ")
print(f"Пароль надежный: {is_strong_password(user_password)}")
```
