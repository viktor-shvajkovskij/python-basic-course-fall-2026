```python
s = "hello"
t = s # t - это просто второй ярлык для того же объекта

# Сравниваем значения
print(f"s == t: {s == t}")
# Вывод: True, потому что значения "hello" и "hello" равны

# Сравниваем идентичность объектов
print(f"s is t: {s is t}")
# Вывод: True, потому что s и t - имена одного и того же объекта
```

```python
# .strip() создает НОВЫЙ объект-строку
s1 = "   hello   ".strip()
s2 = "hello"

# Сравниваем значения
print(f"s1 == s2: {s1 == s2}")
# Вывод: True, потому что значения "hello" и "hello" равны

# Сравниваем идентичность объектов
print(f"s1 is s2: {s1 is s2}")
# Вывод: False, потому что это два разных объекта в памяти!

print(f"ID s1: {id(s1)}")
print(f"ID s2: {id(s2)}") # ID разные!
```

```python
s = "   Hello World!   "

# .upper() -> новая строка в верхнем регистре
upper_string = s.upper() # "   HELLO WORLD!   "

# .lower() -> новая строка в нижнем регистре
lower_string = s.lower() # "   hello world!   "

# .strip() -> новая строка без лишних пробелов по краям
clean_string = s.strip() # "Hello World!"
```

```python
s = "   Hello World!   "
upper_string = s.upper()
lower_string = s.lower()
clean_string = s.strip()

print(f"Оригинал: '{s}'")
print(f"ID оригинала: {id(s)}")
print(f"ID upper_string: {id(upper_string)}") # Другой ID
print(f"ID lower_string: {id(lower_string)}") # Другой ID
print(f"ID clean_string: {id(clean_string)}") # Другой ID
```

```python
# str -> bytes (записываем в файл)
text = "привет"
encoded = text.encode('utf-8')
print(encoded) # b'\xd0\xbf\xd1\x80...'

# bytes -> str (читаем из файла)
decoded = encoded.decode('utf-8')
print(decoded) # 'привет'
```

```python
http_status = 404

if http_status == 200:
    print("OK")
elif http_status == 404:
    print("Not Found")
elif http_status == 500:
    print("Server Error")
else:
    print("Unknown Status")
```

```python
http_status = 404

match http_status:
    case 200:
        print("OK")
    case 404:
        print("Not Found")
    case 500:
        print("Server Error")
    case _: # "Шаблон-джокер"
        print("Unknown Status")
```

```python
# Гораздо чище, чем if status == 401 or status == 403 or ...
match http_status:
    case 200 | 201:
        print("Успех")
    case 401 | 403 | 404:
        print("Ошибка на стороне клиента")
    case 500 | 503:
        print("Ошибка на стороне сервера")
```

```python
import sys

my_obj = "a string"
# Обратите внимание: getrefcount() сам создает временную ссылку,
# поэтому результат всегда на 1 больше, чем мы ожидаем.
print(sys.getrefcount(my_obj)) # Вывод: 2

another_ref = my_obj
print(sys.getrefcount(my_obj)) # Вывод: 3

del another_ref
print(sys.getrefcount(my_obj)) # Вывод: 2
```
