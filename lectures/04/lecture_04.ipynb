```python
def withdraw(balance, amount):
    if amount <= 0:
        raise ValueError("Amount must be positive")
    if amount > balance:
        raise ValueError("Insufficient funds")
    return balance - amount

# Использование
try:
    new_balance = withdraw(100, 150)
except ValueError as e:
    print(f"Transaction failed: {e}")

# Повторный выброс исключения
try:
    process_file()
except PermissionError:
    print("Logging error...")
    raise  # Пробрасываем исключение дальше
```

```python
# Собственное исключение
class InsufficientFundsError(Exception):
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(
            f"Cannot withdraw {amount}. Balance: {balance}"
        )

class BankAccount:
    def __init__(self, balance):
        self.balance = balance
    
    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientFundsError(
                self.balance, amount
            )
        self.balance -= amount

# Использование
try:
    account = BankAccount(100)
    account.withdraw(150)
except InsufficientFundsError as e:
    print(f"Error: {e}")
    print(f"Available: {e.balance}")
```

```python
# Функция с обработкой исключений
def save_user_data(filename, user_data):
    """Save user data to JSON file."""
    try:
        with open(filename, 'w', encoding='utf-8') as f:
            json.dump(user_data, f, indent=2)
    except PermissionError:
        raise PermissionError(f"No permission to write {filename}")
    except Exception as e:
        raise IOError(f"Failed to save data: {e}")

# Тест
def test_save_user_data(tmp_path):
    file = tmp_path / "user.json"
    user = {"name": "Alice", "age": 25}
    
    save_user_data(str(file), user)
    
    assert file.exists()
    with open(file, 'r', encoding='utf-8') as f:
        loaded = json.load(f)
    assert loaded == user
```
