```python
class Car:
    """Класс для описания автомобиля"""
    
    def __init__(self, brand, color, year):
        self.brand = brand      # атрибут
        self.color = color      # атрибут
        self.year = year        # атрибут
        self.speed = 0          # атрибут с начальным значением
```

```python
class Car:
    def __init__(self, brand, color, year):
        self.brand = brand
        self.color = color
        self.year = year
        self.speed = 0
    
    def accelerate(self, amount):      # метод
        self.speed += amount
        print(f"Скорость увеличена на {amount}")
    
    def brake(self):                   # метод
        self.speed = 0
        print("Автомобиль остановлен")
    
    def get_info(self):                # метод
        return f"{self.brand} {self.year} года"

# Использование методов
car1.accelerate(50)
car1.brake()
car1.get_info()
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self._owner = owner              # защищенный атрибут
        self.__balance = balance         # приватный атрибут
    
    def deposit(self, amount):           # публичный метод
        if amount > 0:
            self.__balance += amount
    
    def get_balance(self):               # публичный метод
        return self.__balance
```

```python
class Vehicle:                          # базовый класс
    def __init__(self, brand):
        self.brand = brand
    
    def drive(self):
        print(f"{self.brand} едет")

class Car(Vehicle):                     # производный класс
    def __init__(self, brand, doors):
        super().__init__(brand)         # вызов конструктора родителя
        self.doors = doors
    
    def honk(self):                     # новый метод
        print("Бип-бип!")

car = Car("BMW", 4)
car.drive()                             # наследованный метод
car.honk()                              # собственный метод
```

```python
import math
from abc import ABC, abstractmethod

class Shape(ABC):                       # абстрактный класс
    @abstractmethod
    def area(self):                     # абстрактный метод
        pass
    
    @abstractmethod
    def perimeter(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return math.pi * self.radius ** 2
    
    def perimeter(self):
        return 2 * math.pi * self.radius

# Shape() — ошибка! Абстрактный класс нельзя инстанцировать
circle = Circle(5)
print(circle.area())
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __str__(self):                  # строковое представление
        return f"{self.name}, {self.age} лет"
    
    def __repr__(self):                 # для отладки
        return f"Person('{self.name}', {self.age})"
    
    def __len__(self):
        return self.age
    
    def __eq__(self, other):            # сравнение
        return self.age == other.age

person = Person("Alice", 30)
print(person)                           # Alice, 30 лет
print(len(person))                      # 30
```

```python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius
    
    @property                           # геттер
    def celsius(self):
        return self._celsius
    
    @celsius.setter                     # сеттер
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Невозможная температура")
        self._celsius = value
    
    @property
    def fahrenheit(self):
        return self._celsius * 9/5 + 32

temp = Temperature(25)
print(temp.celsius)                     # 25
temp.celsius = 30                       # использует сеттер
print(temp.fahrenheit)                  # 86.0
```

```python
from abc import ABC, abstractmethod
from datetime import datetime

class Pet(ABC):
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.created_at = datetime.now()
    
    @abstractmethod
    def speak(self):
        pass
    
    def info(self):
        return f"{self.name}, {self.age} лет"

class Dog(Pet):
    def speak(self):
        return "Гав!"

class Cat(Pet):
    def speak(self):
        return "Мяу!"

class PetShelter:
    def __init__(self):
        self.pets = []
    
    def add_pet(self, pet):
        self.pets.append(pet)
    
    def make_all_speak(self):
        for pet in self.pets:
            print(f"{pet.name}: {pet.speak()}")

shelter = PetShelter()
shelter.add_pet(Dog("Rex", 3))
shelter.add_pet(Cat("Whiskers", 2))
shelter.make_all_speak()
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    
    def get_annual_salary(self):
        return self.salary * 12

class Manager(Employee):
    def __init__(self, name, salary, team_size):
        super().__init__(name, salary)
        self.team_size = team_size
    
    def get_annual_salary(self):
        base = super().get_annual_salary()
        bonus = base * 0.2 * (self.team_size / 10)
        return base + bonus

class Developer(Employee):
    def __init__(self, name, salary, languages):
        super().__init__(name, salary)
        self.languages = languages
    
    def get_annual_salary(self):
        base = super().get_annual_salary()
        return base + (len(self.languages) * 1000)

employees = [
    Manager("John", 5000, 5),
    Developer("Alice", 4000, ["Python", "Java", "JS"])
]

for emp in employees:
    print(f"{emp.name}: {emp.get_annual_salary()} в год")
```

```python
class Calculator:
    """Простой калькулятор для базовых операций.
    
    Attributes:
        history (list): История операций
    """
    
    def __init__(self):
        """Инициализирует новый калькулятор."""
        self.history = []
    
    def add(self, a, b):
        """Складывает два числа.
        
        Args:
            a (float): Первое число
            b (float): Второе число
        
        Returns:
            float: Сумма чисел
        
        Examples:
            >>> calc = Calculator()
            >>> calc.add(2, 3)
            5
        """
        result = a + b
        self.history.append(f"{a} + {b} = {result}")
        return result

help(Calculator.add)
```

```python
class Singleton:
    _instance = None
    
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

singleton1 = Singleton()
singleton2 = Singleton()
print(singleton1 is singleton2)         # True

# Или через декоратор:
def singleton(cls):
    instances = {}
    def get_instance(*args, **kwargs):
        if cls not in instances:
            instances[cls] = cls(*args, **kwargs)
        return instances[cls]
    return get_instance

@singleton
class Database:
    def __init__(self):
        self.connection = None
```

```python
class Dog:
    def speak(self):
        return "Гав!"

class Cat:
    def speak(self):
        return "Мяу!"

class AnimalFactory:
    @staticmethod
    def create_animal(animal_type):
        if animal_type == "dog":
            return Dog()
        elif animal_type == "cat":
            return Cat()
        else:
            raise ValueError("Неизвестный тип животного")

# Использование
animal = AnimalFactory.create_animal("dog")
print(animal.speak())
```

```python
class Subject:
    def __init__(self):
        self._observers = []
    
    def attach(self, observer):
        self._observers.append(observer)
    
    def notify(self):
        for observer in self._observers:
            observer.update(self)

class WeatherStation(Subject):
    def __init__(self):
        super().__init__()
        self._temperature = 0
    
    @property
    def temperature(self):
        return self._temperature
    
    @temperature.setter
    def temperature(self, temp):
        self._temperature = temp
        self.notify()

class TemperatureDisplay:
    def update(self, subject):
        print(f"Новая температура: {subject.temperature}°C")

station = WeatherStation()
display = TemperatureDisplay()
station.attach(display)
station.temperature = 25
```

```python
import unittest

class Calculator:
    def add(self, a, b):
        return a + b
    
    def divide(self, a, b):
        if b == 0:
            raise ValueError("Деление на ноль")
        return a / b

class TestCalculator(unittest.TestCase):
    def setUp(self):
        self.calc = Calculator()
    
    def test_add(self):
        self.assertEqual(self.calc.add(2, 3), 5)
    
    def test_divide_by_zero(self):
        with self.assertRaises(ValueError):
            self.calc.divide(10, 0)

if __name__ == '__main__':
    unittest.main()
```
