```python
def say_hello(name):
    return f"Hello, {name}"

# 1. Присваиваем функцию переменной
greet = say_hello
print(greet("World"))

# 2. Передаем функцию как аргумент
def process_greeting(greeter_func, name):
    print(greeter_func(name))

process_greeting(say_hello, "Alice")
```

```python
def create_adder(x):
    # x - 'свободная переменная' для adder
    def adder(y):
        return x + y
    return adder

add_5 = create_adder(5)
print(add_5(10))

add_10 = create_adder(10)
print(add_10(10))
```

```python
def slow_function():
    import time
    time.sleep(2)
    return "Done!"
```

```python
import time
def timer_decorator(func):
    def wrapper():
        start_time = time.time()
        result = func() # Вызываем оригинальную функцию
        end_time = time.time()
        print(f"Функция {func.__name__} выполнялась {end_time - start_time:.2f} с.")
        return result
    return wrapper
```

```python
timed_slow_function = timer_decorator(slow_function)
timed_slow_function()
```

```python
def timer_decorator(func):
    def wrapper(*args, **kwargs): # Принимаем любые аргументы
        start_time = time.time()
        result = func(*args, **kwargs) # Передаем их дальше
        end_time = time.time()
        print(f"Функция {func.__name__} выполнялась {end_time - start_time:.2f} с.")
        return result
    return wrapper

@timer_decorator
def slow_function_with_args(delay, message):
    time.sleep(delay)
    return f"Done with message: {message}"

slow_function_with_args(1, "Important task")

print(slow_function_with_args.__name__)
```

```python
import functools

def timer_decorator(func):
    @functools.wraps(func) # Копируем метаданные
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"Функция {func.__name__} выполнялась {end_time - start_time:.2f} с.")
        return result
    return wrapper

@timer_decorator
def slow_function_with_args(delay, message):
    time.sleep(delay)
    return f"Done with message: {message}"

print(slow_function_with_args.__name__)
```

```python
class Fibonacci:
    def __init__(self, limit):
        self.limit = limit
        self.a, self.b = 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.a > self.limit:
            raise StopIteration
        result = self.a
        self.a, self.b = self.b, self.a + self.b
        return result

# Как использовать:
fib_iterator = Fibonacci(100)
for num in fib_iterator:
    print(num)
```

```python
def fibonacci_generator(limit):
    a, b = 0, 1
    while a <= limit:
        yield a
        a, b = b, a + b

# Как использовать:
for num in fibonacci_generator(100):
    print(num)
```

```python
# Классический цикл for
squares = []
for x in range(10):
    if x % 2 == 0:
        squares.append(x * x)

# Эквивалент с помощью list comprehension
squares_comp = [x * x for x in range(10) if x % 2 == 0]

print(squares_comp) # [0, 4, 16, 36, 64]
```
