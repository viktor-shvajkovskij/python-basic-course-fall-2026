### Практический пример: RGB Манипуляции с изображением
Загружаем реальное изображение и работаем с цветовыми каналами.
Это наглядно показывает, что любое изображение — это просто трехмерный массив чисел.

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

try:
    img = mpimg.imread('1.png')
except FileNotFoundError:
    img = np.random.rand(100, 100, 3)

print(f"Размер изображения: {img.shape}")

tinted = img.copy()

tinted[:, :, 1] = 0 # Зануляем Green
tinted[:, :, 2] = 0 # Зануляем Blue


negative = 1 - img

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.title("Original")
plt.imshow(img)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Red Channel Only")
plt.imshow(tinted)
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Negative")
plt.imshow(negative)
plt.axis('off')

plt.show()
```

### Практический пример: Генерация случайного блуждания
Симуляция движения частицы (или цены акции), которая на каждом шаге делает шаг +1 или -1.

```python
import numpy as np
import matplotlib.pyplot as plt
steps = 1000
# Генерируем 1000 шагов: либо -1, либо 1
draws = np.random.choice([-1, 1], size=steps)

# Считаем кумулятивную сумму (траекторию)
walk = draws.cumsum()

plt.plot(walk)
plt.title("Случайное блуждание (Random Walk)")
plt.xlabel("Шаги")
plt.ylabel("Позиция")
plt.show()
```

### Практический пример: Анализ зашумленного сигнала

```python
x = np.linspace(0, 10, 100)
pure_signal = np.sin(x)

# Добавляем шум (случайные числа с нормальным распределением)
noise = np.random.normal(0, 0.2, 100)
noisy_signal = pure_signal + noise

# Простой фильтр (скользящее среднее)
# Свертка сигнала с окном из 5 единиц, деленная на 5
filtered_signal = np.convolve(noisy_signal, np.ones(5)/5, mode='same')

plt.figure(figsize=(10, 6))
plt.plot(x, pure_signal, label='Чистый сигнал', linewidth=2, linestyle='--')
plt.plot(x, noisy_signal, label='Зашумленный сигнал', alpha=0.5)
plt.plot(x, filtered_signal, label='Отфильтрованный', linewidth=2)
plt.legend()
plt.title("Фильтрация сигнала")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Создаем синий фон (100x100)
# Синий цвет в RGB: [0, 0, 255]
image_rgb = np.zeros((100, 100, 3), dtype=np.uint8)
image_rgb[:, :] = [0, 0, 255] # Заливаем синим

# 2. Рисуем красный квадрат в центре
# Красный цвет: [255, 0, 0]
image_rgb[25:75, 25:75] = [255, 0, 0]

# 3. Делаем негатив
negative_rgb = 255 - image_rgb

# 4. Вывод
plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.title("Оригинал (Красный на Синем)")
plt.imshow(image_rgb)

plt.subplot(1, 2, 2)
plt.title("Негатив (Голубой на Желтом)")
# Синий (0,0,255) -> Желтый (255,255,0)
# Красный (255,0,0) -> Голубой (0,255,255)
plt.imshow(negative_rgb)

plt.show()
```
