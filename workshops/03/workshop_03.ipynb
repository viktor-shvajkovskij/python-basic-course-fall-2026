```python
import copy

# Создаем оригинальную вложенную структуру
matrix_a = [ [1, 2, 3], [4, 5, 6] ]

# 1. Поверхностная копия (shallow copy)
matrix_b = matrix_a.copy() [:]

# 2. Глубокая копия (deep copy)
matrix_c = copy.deepcopy(matrix_a)

matrix_b[0][0] = 99

print(f"Matrix A (Original): {matrix_a}")
print(f"Matrix B (Shallow):  {matrix_b}")
print(f"Matrix C (Deep):     {matrix_c}")
```

```python
A = {1, 2, 3}; B = {3, 4}

A | B    # {1, 2, 3, 4}   объединение      A.union(B)
A & B    # {3}            пересечение      A.intersection(B)
A - B    # {1, 2}         разность         A.difference(B)
A ^ B    # {1, 2, 4}      симметр. разность A.symmetric_difference(B)

A <= B   # подмножество (нестрогое)        A.issubset(B)
A <  B   # строгое подмножество
A >= B   # надмножество                    A.issuperset(B)
A.isdisjoint(B)   # нет общих элементов
```

### Словари (dictionaries)

```python
data = {'a': 1, 'b': 2}

keys = data.keys()
print(f"Initial keys: {keys}")

data['c'] = 3

print(f"Final keys: {keys}")
```

```python
a = {'x': 1, 'y': 2}
b = {'y': 9, 'z': 3}

print(a.keys() & b.keys())   # {'y'}  — общие ключи
print(a.keys() - b.keys())   # {'x'}  — есть только в a
print(a.keys() | b.keys())   # {'x', 'y', 'z'}
print(a.items() & b.items()) # set()  — одинаковых пар нет: y=2 против y=9
```

```python
data = {}
words = ['apple', 'banana', 'apple', 'cherry']

for word in words:
    if word not in data:
        data[word] = 0
    data[word] += 1

print(f"Word counts: {data}")
```

```python
data = {}
words = ['apple', 'banana', 'apple', 'cherry']

for word in words:
    data[word] = data.setdefault(word, 0) + 1

print(data)
```

```python
import ctypes

class DynamicArray:

    def __init__(self):
        self._capacity = 8
        self._n = 0
        self._A = self._make_array(self._capacity)

    def _make_array(self, c):
        return (c * ctypes.py_object)()
    
    def __len__(self):
        return self._n
    
    def __getitem__(self, i):
        if not 0 <= i < self._n:
            raise IndexError('invalid index')
        return self._A[i]
    
    def append(self, obj):
        if self._n == self._capacity:
            self._resize(self._capacity * 2)

        self._A[self._n] = obj
        self._n += 1

    def _resize(self, new_capacity):
        B = self._make_array(new_capacity)
        for i in range(self._n):
            B[i] = self._A[i]
        self._A = B
        self._capacity = new_capacity

    
    
dynamic_array = DynamicArray()
default_list = []
```

```python
import timeit

time_taken = timeit.timeit(lambda: [dynamic_array.append(i) for i in range(100000)], number=1)
time_taken_list = timeit.timeit(lambda: [default_list.append(i) for i in range(100000)], number=1)
print(time_taken)
print(time_taken_list)
```

```python
class HashTable:
    """
    Упрощенная, но функциональная реализация словаря (хэш-таблицы).
    
    Поддерживает:
    - Добавление/обновление элементов (`my_dict[key] = value`)
    - Получение элементов (`my_dict[key]`)
    - Разрешение коллизий методом цепочек.
    - Динамическое расширение при достижении порога заполнения.
    """
    
    # Порог заполнения. Если size/capacity > 0.67, мы удваиваем размер.
    MAX_LOAD_FACTOR = 2/3 
    
    def __init__(self, capacity=8):
        """
        Инициализирует хэш-таблицу с начальной вместимостью.
        """
        self._capacity = capacity
        self._buckets = [[] for _ in range(self._capacity)]
        self._size = 0

    def _hash(self, key):
        """
        Вычисляет хэш ключа и приводит его к индексу в диапазоне [0, capacity-1].
        """
        return hash(key) % self._capacity

    def __setitem__(self, key, value):
        """
        Добавляет пару (ключ, значение) или обновляет значение, если ключ уже существует.
        """
        # 1. Проверяем, не нужно ли расширить таблицу ПЕРЕД добавлением.
        if (self._size + 1) / self._capacity > self.MAX_LOAD_FACTOR:
            self._resize(2 * self._capacity)
        
        index = self._hash(key)
        bucket = self._buckets[index]
        
        # 2. Ищем ключ в существующей цепочке для обновления.
        for i, (k, v) in enumerate(bucket):
            if k == key:
                bucket[i] = (key, value)
                return # Обновили и вышли
        
        # 3. Если ключ не найден, добавляем новую пару в цепочку.
        bucket.append((key, value))
        self._size += 1

    def __getitem__(self, key):
        """
        Возвращает значение по ключу. Вызывает KeyError, если ключ не найден.
        """
        index = self._hash(key)
        bucket = self._buckets[index]
        
        # Ищем ключ в цепочке.
        for k, v in bucket:
            if k == key:
                return v # Нашли
        
        # Если цикл завершился, а ключ не найден.
        raise KeyError(f"Key not found: {key}")

    def _resize(self, new_capacity):
        """
        Создает новую, большую таблицу и переносит в нее все старые элементы
        с пересчетом их хэшей для новых бакетов.
        """
        # Сохраняем ссылку на старые данные.
        old_buckets = self._buckets

        # Создаем лист бакетов с новой вместимостью.
        self._capacity = new_capacity
        self._buckets = [[] for _ in range(self._capacity)]
        self._size = 0  # Размер будет восстановлен при добавлении.

        # Проходим по всем элементам в старой таблице и добавляем их в новую.
        # Метод __setitem__ сам вычислит правильный новый индекс для каждого элемента.
        for bucket in old_buckets:
            for key, value in bucket:
                self.__setitem__(key, value)

    def __len__(self):
        """Возвращает количество элементов в таблице."""
        return self._size
        
    def __str__(self):
        """Для красивого вывода на печать."""
        items = []
        for bucket in self._buckets:
            # bucket - это список пар, например [(1, 'a'), (11, 'b')]
            for k, v in bucket:
                items.append(f"{repr(k)}: {repr(v)}")
        return "{" + ", ".join(items) + "}"
```
